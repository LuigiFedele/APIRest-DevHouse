```python
import setuptools
import os
from pyspark.sql import SparkSession

os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-17-openjdk-amd64"
os.environ["PATH"] += os.pathsep + os.path.join(os.environ["JAVA_HOME"], "bin")


appName = "PySpark Trabalho de Big Data"
master = "local"

spark = SparkSession.builder.appName(appName).master(master).getOrCreate()
```

```python
imdb_df = spark.read.csv('imdb-reviews-pt-br.csv',
                         header=True,
                         quote="\"",
                         escape="\"",
                         encoding="UTF-8")
```

# **Questão 1**

```python
def map1(x):
    return (x[3], int(x[0]))
```

```python
def reduceByKey1(x, y):
    return x + y
```

```python
resultado_q1 = imdb_df.rdd.map(map1).reduceByKey(reduceByKey1).collect()
print("Luigi Francesco Fedele - RU 4517608")
print("Resultado Questão 1:", resultado_q1)
```

# **Questão 2**

```python
def map2(x):
    # x[1] é text_en, x[2] é text_pt, x[3] é sentiment
    palavras_en = len(x[1].split()) if x[1] else 0
    palavras_pt = len(x[2].split()) if x[2] else 0
    return (x[3], (palavras_en, palavras_pt))
```

```python
def reduceByKey2(x, y):
    return (x[0] + y[0], x[1] + y[1])
```

```python
resultado_q2 = imdb_df.rdd.map(map2).reduceByKey(reduceByKey2).collect()
print("Resultado Questão 2:", resultado_q2)
```

```python
for sentiment, (total_en, total_pt) in resultado_q2:
    if sentiment == 'neg':
        diferenca = total_pt - total_en
        print(f"Total palavras Inglês (negativo): {total_en}")
        print(f"Total palavras Português (negativo): {total_pt}")
        print(f"Diferença (Português - Inglês): {diferenca}")
```
